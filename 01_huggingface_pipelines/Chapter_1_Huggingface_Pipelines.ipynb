### 2. Text Classification

```python
classifier = pipeline("text-classification")

result = classifier("The tech industry saw significant growth this quarter with major players announcing record profits.")
print("Text Classification Result:", result)


classifier = pipeline("text-classification", model="textattack/bert-base-uncased-ag-news")
result = classifier("The tech industry saw significant growth this quarter with major players announcing record profits.")

print("Text Classification Result:", result)
```

### 3. NER

```python
from transformers import pipeline

ner = pipeline("ner")
result = ner("Apple Inc. announced that its CEO Tim Cook visited London on 12th August 2024 to discuss new opportunities. They love Manali")
print(result)
print()

advanced_ner = pipeline( "ner", model="dbmdz/bert-large-cased-finetuned-conll03-english")
results = advanced_ner("Apple Inc. announced that its CEO Tim Cook visited London on 12th August 2024 to discuss new opportunities. They love Manali")
print(results)
```

### 4. Qn Answering

```python
from transformers import pipeline


qa_input = {"question": "What is the capital of England?", "context": "France is a country in Europe. Its capital is Paris."}

qa = pipeline( "question-answering")
qa_advanced = pipeline( "question-answering", model="deepset/roberta-large-squad2")

result = qa(qa_input)
print(result)

result = qa_advanced(qa_input)
print(result)
```

### 5. Summarization

```python
from transformers import pipeline

summarizer = pipeline( "summarization")
summarizer_advanced = pipeline( "summarization", model="facebook/bart-large-cnn")

text = """
Climate change is one of the most pressing global issues facing humanity. It has led to a myriad of environmental, social, and economic consequences that affect communities worldwide. The global temperature has been steadily rising over the past century, leading to unprecedented changes in weather patterns. Sea levels are rising due to the melting of glaciers and polar ice, putting coastal regions and small island nations at risk.

Extreme weather events, including hurricanes, droughts, and wildfires, have become more frequent and intense, causing widespread devastation. In addition to these environmental impacts, climate change is influencing agriculture, water supplies, and public health. Farmers are finding it increasingly difficult to predict growing seasons and manage crop production, leading to food shortages and increased prices. Water resources are under stress, with some regions experiencing severe shortages, while others deal with flooding.

Public health is also at risk, as changes in climate can exacerbate the spread of infectious diseases and lead to heat-related illnesses. Governments and organizations around the world are taking steps to address these challenges by reducing greenhouse gas emissions, investing in renewable energy sources, and implementing policies aimed at sustainability. Efforts to combat climate change require a coordinated global response and a long-term commitment to preserving the planet for future generations.
"""


result = summarizer_advanced(text)
print(result)
print()

result = summarizer(text)
print(result)
```

### 6. Translation

```python
from transformers import pipeline

translator = pipeline("translation_en_to_fr")
translator_adv = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")


result = translator("The quick brown fox jumps over the lazy dog.")
print(result)

translation = translator_adv("The quick brown fox jumps over the lazy dog.")
print("Translation Result:", translation)
```

### 7. Text Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2", truncation=True, pad_token_id=50256)
# generator_advanced = pipeline("text-generation", model="EleutherAI/gpt-neo-2.7B", truncation=True, pad_token_id=50256)
# generator_advanced = pipeline("text-generation", model="EleutherAI/gpt-neo-1.3B", truncation=True, pad_token_id=50256)
generator_advanced = pipeline("text-generation", model="EleutherAI/gpt-neo-125M", truncation=True, pad_token_id=50256)

input_text = "The future of AI is"


result_default = generator(input_text, max_length=500, num_return_sequences=1, truncation=True)
print("Default Text Generation Result:")
print(result_default)


result_advanced = generator_advanced(input_text, max_length=500, num_return_sequences=1, truncation=True)
print("Advanced Text Generation Result:")
print(result_advanced)
print()
```

### 8. Fill-Mask

```python
from transformers import pipeline

unmasker = pipeline("fill-mask")
result = unmasker("Transformers are <mask> in AI.", top_k=2)

print()
print(result)
```

### 9. Zero Shot Learning

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")
result = classifier("What is GenAI?", candidate_labels=["ai", "dsa","backend","frontend","cyber security"])

print()
print(result)
```

```python
import torch
from diffusers import StableDiffusionPipeline

model_id = "CompVis/stable-diffusion-v1-4"
device = "cuda"


pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to(device)

prompt = "a photo of an astronaut riding a horse on mars"
image = pipe(prompt).images[0]

image.save("astronaut_rides_horse.png")
```
